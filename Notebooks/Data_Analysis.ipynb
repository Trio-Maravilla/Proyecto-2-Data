# Analisis 

```python
import pandas as pd
```

```python
datos_completos = pd.read_csv("../Data/Processed/Metricas delitos segun departamento - 2023-2025.csv")
datos_2023 = pd.read_csv("../Data/Processed/Metricas delitos segun departamento - Filtrado 2023.csv")
datos_2024 = pd.read_csv("../Data/Processed/Metricas delitos segun departamento - Filtrado 2024.csv")
datos_2025 = pd.read_csv("../Data/Processed/Metricas delitos segun departamento - Filtrado 2025.csv")
```

## Correlaciones 

### General

```python
excluir = ["DEPARTAMENTO", "ANO", "Población"]
dimensiones_ipm = [c for c in datos_completos.columns
                   if c not in excluir and not c.startswith(("CANTIDAD_", "TASA_"))]
cols_tasas = [c for c in datos_completos.columns if c.startswith("TASA_")]
df_depto = (datos_completos
            .groupby("DEPARTAMENTO")[cols_tasas + dimensiones_ipm]
            .mean())

matriz_depto = (df_depto
                .corr(method="spearman")
                .loc[cols_tasas, dimensiones_ipm])
matriz_depto.round(2)


# añadimos las tasas
```
