## __0.0 CARGA DE DATOS:__

```python
import pandas as pd
from pathlib import Path # Librería a utilizar para enviar los datos ya procesados

pd.options.display.max_seq_items = None # Para quitar el límite de los prints()

carpeta_salida_csv = Path("../Data/Processed") # Carpeta de salida del data/processed
carpeta_salida_csv.mkdir(parents=True, exist_ok=True)

del_med_ambiente = pd.read_csv("../Data/Raw/DELITOS_CONTRA_EL_MEDIO_AMBIENTE_20261102.csv")
del_informaticos = pd.read_csv("../Data/Raw/DELITOS_INFORMÁTICOS_20261102.csv")
del_sexuales = pd.read_csv("../Data/Raw/DELITOS_SEXUALES_20261102.csv")
extorsion = pd.read_csv("../Data/Raw/EXTORSIÓN_20261102.csv")
hurto_vehiculos = pd.read_csv("../Data/Raw/HURTO_A_VEHÍCULOS_20261102.csv")
hurto_personas = pd.read_csv("../Data/Raw/HURTO_PERSONAS_20261102.csv", dtype={"COD_DEPTO": str, "COD_MUNI": str}) # Acá solía soltar un error de tipos para ambas columnas
incautacion_estupefacientes = pd.read_csv("../Data/Raw/Incautación_de_Estupefacientes._20261102.csv", dtype={"CODIGO DANE": str}) # Misma situación
masacres = pd.read_csv("../Data/Raw/MASACRES_20261102.csv")
pirateria_terrestre = pd.read_csv("../Data/Raw/PIRATERÍA_TERRESTRE_20261102.csv")
secuestro = pd.read_csv("../Data/Raw/SECUESTRO_20261102.csv")
terrorismo = pd.read_csv("../Data/Raw/TERRORISMO_20261102.csv")
violencia_intrafamiliar = pd.read_csv("../Data/Raw/VIOLENCIA_INTRAFAMILIAR_20261102.csv")

datasets = {
    "del_med_ambiente": del_med_ambiente.copy(),
    "del_informaticos": del_informaticos.copy(),
    "del_sexuales": del_sexuales.copy(),
    "extorsion": extorsion.copy(),
    "hurto_vehiculos": hurto_vehiculos.copy(),
    "hurto_personas": hurto_personas.copy(),
    "incautacion_estupefacientes": incautacion_estupefacientes.copy(),
    "masacres": masacres.copy(),
    "pirateria_terrestre": pirateria_terrestre.copy(),
    "secuestro": secuestro.copy(),
    "terrorismo": terrorismo.copy(),
    "violencia_intrafamiliar": violencia_intrafamiliar.copy()
}

df_pobreza_2023 = pd.read_excel("../Data/Raw/IPM_Departamental_2023.xlsx")
df_pobreza_2024 = pd.read_excel("../Data/Raw/IPM_Departamental_2024.xlsx")
df_pobreza_2025 = pd.read_excel("../Data/Raw/IPM_Departamental_2025.xlsx")
```

#### SHAPE DE TODOS LOS DATASETS:

## __1.2 CONVERSIÓN DE FECHAS:__

#### UNIFICACIÓN DE FORMATO DE FECHA PARA CADA DATASET

```python
formato_especial = {
    "del_informaticos": "%Y %b %d %I:%M:%S %p",
    "masacres":         "%Y %b %d %I:%M:%S %p",
    "extorsion":        "%m/%d/%Y",   # mes/día/año
}

for nombre, df in datasets.items():
    if nombre in formato_especial:
        df["FECHA HECHO"] = pd.to_datetime(
            df["FECHA HECHO"],
            format=formato_especial[nombre],
            errors="coerce",
        )
    else:
        df["FECHA HECHO"] = pd.to_datetime(
            df["FECHA HECHO"],
            format="mixed",
            errors="coerce",
        )

    # Verificación: fechas que no se pudieron parsear
    print(f"{nombre}: {df['FECHA HECHO'].isna().sum()} NaT")
```

---

## __1.3 NORMALIZACIÓN DE DEPARTAMENTOS:__

#### NORMALIZACIÓN DE NOMBRES PARA CADA DEPARTAMENTO DE CADA DATASET

```python
for nombre, df in datasets.items():
    df["DEPARTAMENTO"] = (df["DEPARTAMENTO"]
                          .str.replace("Á", "A")
                          .str.replace("É", "E")
                          .str.replace("Í", "I")
                          .str.replace("Ó", "O")
                          .str.replace("Ú", "U")
                          .str.replace("SAN ANDRES ISLAS", "SAN ANDRES")
                          .str.replace("VALLE", "VALLE DEL CAUCA")
                          .str.replace("BOGOTA D.C.", "BOGOTA")
                          .str.replace("LA GUAJIRA", "GUAJIRA")
                          .str.replace("VALLE DEL CAUCA DEL CAUCA", "VALLE DEL CAUCA")
                        )
```

---

## __1.4 CREACIÓN DE COLUMNA AÑO Y FILTRADO 2023-2025:__

#### CREAR LA COLUMNA AÑO 

```python
for nombre, df in datasets.items():
    df["ANO"] = df["FECHA HECHO"].dt.year
```

#### (PASO INTERMEDIO): REVISAR RANGOS MÍNIMOS Y MÁXIMOS DE AÑOS PARA CADA DATASET JUNTO A SU SHAPE ANTES DE APLICAR EL FILTRO

```python
for nombre, df in datasets.items():
    print(f"\nDataframe: {nombre}, shape: {df.shape}")
    print("Año mínimo:", df["ANO"].min())
    print("Año máximo:", df["ANO"].max())
    print("="*50)
```

#### APLICAR FILTRO 2023-2025

```python
for nombre, df in datasets.items():
    datasets[nombre] = df[(df["ANO"] >= 2023) & (df["ANO"] <= 2025)]
```

#### REVISAR EN CUÁNTAS FILAS DISMINUYERON LOS DATASETS LUEGO DE APLICAR EL FILTRO

```python
for nombre, df in datasets.items():
    print(f"{nombre}: {df.shape}") 
```

## __1.5 NULOS Y DUPLICADOS:__

#### SUMA DE NULOS POR CADA COLUMNA DE CADA DATASET

```python
for nombre, df in datasets.items():
    print(f"\nDataset: {nombre}")
    print(df.isnull().sum().sort_values(ascending=False).to_string())
```

#### HALLAZGOS:

3. El dataset de masacres tiene 277 valores nulos para la columna ZONA. La cantidad de filas que tiee el dataframe es de 277, por esta razón intuyo que sencillamente no se implementó conocer la zona exacta de la masacre y decido eliminar la columna por completo. 

#### ELIMINACIÓN DE NULOS

```python
datasets["del_sexuales"].dropna(subset=["SEXO"], inplace=True) 
print(f"Shape de delitos sexuales luego de hacer dropna: {datasets["del_sexuales"].shape}")

datasets["masacres"].drop(columns=["ZONA"], inplace=True, errors="ignore")
print(f"Shape de masacres luego de eliminar columna ZONA: {datasets["masacres"].shape}")
```

#### AVERIGUAR LA SUMA DE DUPLICADOS POR CADA DATAFRAME

```python
for nombre, df in datasets.items():
    print(f"\nDataframe: {nombre}, shape: {df.shape}")
    print("Número de duplicados:",df.duplicated(keep="first").sum()) # El parámetro keep="first" indica que se mantendrá la primera aparición de cada fila duplicada y se marcarán como duplicadas las siguientes apariciones.
    print("="*50)
```

#### HALLAZGOS / DESICIÓN:
Para este punto de la limpieza de datos decidimos dejar los registros intactos en cuanto a duplicados debido a que el hecho de que exista un duplicado no representa automaticamente un error, lo que representa es que en el mismo día y en el mismo lugar sucedieron N cantidad de casos que pandas toma como duplicados debido a la falta de una columna que sirva para  como identificador único.

## __1.6 MERGE DE DATAFRAMES (CRÍMENES):__

#### PASO PREVIO IMPORTANTE: AGRUPACIÓN DE DATASETS x AÑO y DEPARTAMENTO SOLO CON LA COLUMNA "CANTIDAD" DE CASOS

```python
agregados = {}
for nombre, df in datasets.items():
    d = df.copy()
    if nombre == "masacres":
        d = d.rename(columns={"CASOS": "CANTIDAD"})
    if nombre == "incautacion_estupefacientes":
        d["CANTIDAD"] = 1  
    d = d.groupby(["DEPARTAMENTO", "ANO"], as_index=False)["CANTIDAD"].sum()
    agregados[nombre] = d.rename(columns={"CANTIDAD": f"CANTIDAD_{nombre}"})

for nombre, df in agregados.items():
    print(nombre, df.shape)
```

#### MERGE FINAL ENTRE TODOS LOS DATASETS DE CRÍMENES

```python
claves = ["DEPARTAMENTO", "ANO"]

df_crimenes = agregados["del_med_ambiente"]
for nombre, df in agregados.items():
    if nombre != "del_med_ambiente":
        df_crimenes = df_crimenes.merge(df, on=claves, how="outer")
```

#### VERIFICACIÓN

```python
print(df_crimenes.shape) 
print(df_crimenes.isna().sum()[lambda s: s > 0])
```

# __PUNTO 2: TRATAMIENTO Y MERGE DE DATASETS IPM__

---

## __2.0 CARGA DE DATOS:__

```python
df_pobreza_2023 = pd.read_excel("../Data/Raw/IPM_Departamental_2023.xlsx")
df_pobreza_2024 = pd.read_excel("../Data/Raw/IPM_Departamental_2024.xlsx")
df_pobreza_2025 = pd.read_excel("../Data/Raw/IPM_Departamental_2025.xlsx")
```

---

## __2.1 UNIÓN DE DATASETS DE POBREZA__

#### Unimos los datasets de pobreza en 1 y arreglamos formato departamentos

```python
df_pobreza_2023['ANO'] = 2023
df_pobreza_2024['ANO'] = 2024
df_pobreza_2025['ANO'] = 2025
df_pobreza = pd.concat([df_pobreza_2023, df_pobreza_2024, df_pobreza_2025], ignore_index=True)
df_pobreza["DEPARTAMENTO"] = (df_pobreza["DEPARTAMENTO"]
                          .str.upper()
                          .str.replace("Á", "A")
                          .str.replace("É", "E")
                          .str.replace("Í", "I")
                          .str.replace("Ó", "O")
                          .str.replace("Ú", "U")
                          .str.replace("LA GUAJIRA", "GUAJIRA")
)
```

#### Revisamos la unión

```python
df_pobreza
```

---

## __2.2 MERGE ENTRE DATAFRAMES DE CRÍMENES Y DE POBREZA UNIFICADOS__

```python
df_final = pd.merge(
    df_crimenes, 
    df_pobreza, 
    on=['DEPARTAMENTO', 'ANO'], 
    how='right'
)
```

### Llenamos nulos con 0

```python
cols_delitos = [c for c in df_final.columns if c.startswith("CANTIDAD_")]
df_final[cols_delitos] = df_final[cols_delitos].fillna(0)
print(set(df_crimenes["DEPARTAMENTO"]) == set(df_pobreza["DEPARTAMENTO"])) # Esto es para saber que la unión entre departamentos es correcto para ambos dataframes
```

### EL RESULTADO FINAL DEL MERGE Y DE LA UNIFICACION DE DEPARTAMENTOS ANTERIOR ES EL SIGUIENTE:

```python
print("SHAPE FINAL DEL MERGE:", df_final.shape)
print("NUMERO DE DEPARTAMENTOS (CONTANDO A BOGOTA COMO UNO):", df_final["DEPARTAMENTO"].nunique())
```

#### Print final del df_final para conocer sus valores

```python
df_final
```

---

## __2.3 REALIZAR PEQUEÑA REVISIÓN DE DATOS SOBRE EL NUEVO MERGE__

#### Analisis de correlaciones aquí porque el kernel de python nos odia

```python
# Tasa por 100.000 habitantes
for c in cols_delitos:
    df_final["TASA_" + c.replace("CANTIDAD_", "")] = df_final[c] / df_final["Población"] * 100_000
cols_tasas = [c for c in df_final.columns if c.startswith("TASA_")]
```

#### Revisión de correlaciones de Spearman entre Tasas de crímenes y IPM

```python
dimensiones_ipm = [c for c in df_pobreza.columns if c not in ["DEPARTAMENTO", "ANO", "Población"]]

matriz = (df_final[cols_tasas + dimensiones_ipm]
          .corr(method="spearman")
          .loc[cols_tasas, dimensiones_ipm])

matriz.round(2)
```

### SEPARAR df_final EN TRES (POR CADA AÑO)

```python
df_2023 = df_final[df_final["ANO"] == 2023]
df_2024 = df_final[df_final["ANO"] == 2024]
df_2025 = df_final[df_final["ANO"] == 2025]
```

### ENVIAR LOS NUEVOS CSV A LA CARPETA DE DATA PROCESSED:

```python
df_2023.to_csv(carpeta_salida_csv / "Metricas delitos segun departamento - Filtrado 2023.csv", index=False)
df_2024.to_csv(carpeta_salida_csv / "Metricas delitos segun departamento - Filtrado 2024.csv", index=False)
df_2025.to_csv(carpeta_salida_csv / "Metricas delitos segun departamento - Filtrado 2025.csv", index=False)
```
