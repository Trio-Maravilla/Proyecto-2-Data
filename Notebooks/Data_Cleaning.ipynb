## __1.2 CONVERSIÓN DE FECHAS:__

#### UNIFICACIÓN DE FORMATO DE FECHA PARA CADA DATASET

```python
formato_especial = {
    "del_informaticos": "%Y %b %d %I:%M:%S %p",
    "masacres":         "%Y %b %d %I:%M:%S %p",
    "extorsion":        "%m/%d/%Y",   # mes/día/año
}

for nombre, df in datasets.items():
    if nombre in formato_especial:
        df["FECHA HECHO"] = pd.to_datetime(
            df["FECHA HECHO"],
            format=formato_especial[nombre],
            errors="coerce",
        )
    else:
        df["FECHA HECHO"] = pd.to_datetime(
            df["FECHA HECHO"],
            format="mixed",
            errors="coerce",
        )

    # Verificación: fechas que no se pudieron parsear
    print(f"{nombre}: {df['FECHA HECHO'].isna().sum()} NaT")
```

## __1.6 MERGE DE DATAFRAMES (CRÍMENES):__

#### PASO PREVIO IMPORTANTE: AGRUPACIÓN DE DATASETS x AÑO y DEPARTAMENTO SOLO CON LA COLUMNA "CANTIDAD" DE CASOS

```python
agregados = {}
for nombre, df in datasets.items():
    d = df.copy()
    if nombre == "masacres":
        d = d.rename(columns={"CASOS": "CANTIDAD"})
    d = d.groupby(["DEPARTAMENTO", "ANO"], as_index=False)["CANTIDAD"].sum()
    agregados[nombre] = d.rename(columns={"CANTIDAD": f"CANTIDAD_{nombre}"})

for nombre, df in agregados.items():
    print(nombre, df.shape)
```

#### MERGE FINAL ENTRE TODOS LOS DATASETS DE CRÍMENES

```python
from functools import reduce

claves = ["DEPARTAMENTO", "ANO"]
df_crimenes = reduce(
    lambda izq, der: izq.merge(der, on=claves, how="outer", validate="one_to_one"),
    agregados.values(),
)
```

#### VERIFICACIÓN

```python
print(df_crimenes.shape)   # esperado: (99, 14)
print(df_crimenes.isna().sum()[lambda s: s > 0])
```

### (Nelson) Unimos los datasets de pobreza en 1 y arreglamos formato departamentos

```python
df_pobreza_2023['ANO'] = 2023
df_pobreza_2024['ANO'] = 2024
df_pobreza_2025['ANO'] = 2025
df_pobreza = pd.concat([df_pobreza_2023, df_pobreza_2024, df_pobreza_2025], ignore_index=True)
df_pobreza["DEPARTAMENTO"] = (df_pobreza["DEPARTAMENTO"]
                          .str.upper()
                          .str.replace("Á", "A")
                          .str.replace("É", "E")
                          .str.replace("Í", "I")
                          .str.replace("Ó", "O")
                          .str.replace("Ú", "U")
                          .str.replace("LA GUAJIRA", "GUAJIRA")
)
```

```python
df_final = pd.merge(
    df_crimenes, 
    df_pobreza, 
    on=['DEPARTAMENTO', 'ANO'], 
    how='right'
)
```
