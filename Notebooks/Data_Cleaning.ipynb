### __CARGA DE DATOS:__

```python
import pandas as pd
from pathlib import Path # Librería a utilizar para enviar los datos ya procesados

pd.options.display.max_seq_items = None # Para quitar el límite de los prints()

carpeta_salida_csv = Path("../Data/Processed") # Carpeta de salida del data/processed
carpeta_salida_csv.mkdir(parents=True, exist_ok=True)

del_med_ambiente = pd.read_csv("../Data/Raw/DELITOS_CONTRA_EL_MEDIO_AMBIENTE_20261102.csv")
del_informaticos = pd.read_csv("../Data/Raw/DELITOS_INFORMÁTICOS_20261102.csv")
del_sexuales = pd.read_csv("../Data/Raw/DELITOS_SEXUALES_20261102.csv")
extorsion = pd.read_csv("../Data/Raw/EXTORSIÓN_20261102.csv")
hurto_vehiculos = pd.read_csv("../Data/Raw/HURTO_A_VEHÍCULOS_20261102.csv")
hurto_personas = pd.read_csv("../Data/Raw/HURTO_PERSONAS_20261102.csv")
incautacion_estupefacientes = pd.read_csv("../Data/Raw/Incautación_de_Estupefacientes._20261102.csv")
masacres = pd.read_csv("../Data/Raw/MASACRES_20261102.csv")
pirateria_terrestre = pd.read_csv("../Data/Raw/PIRATERÍA_TERRESTRE_20261102.csv")
secuestro = pd.read_csv("../Data/Raw/SECUESTRO_20261102.csv")
terrorismo = pd.read_csv("../Data/Raw/TERRORISMO_20261102.csv")
violencia_intrafamiliar = pd.read_csv("../Data/Raw/VIOLENCIA_INTRAFAMILIAR_20261102.csv")

datasets = {
    "del_med_ambiente": del_med_ambiente,
    "del_informaticos": del_informaticos,
    "del_sexuales": del_sexuales,
    "extorsion": extorsion,
    "hurto_vehiculos": hurto_vehiculos,
    "hurto_personas": hurto_personas,
    "incautacion_estupefacientes": incautacion_estupefacientes,
    "masacres": masacres,
    "pirateria_terrestre": pirateria_terrestre,
    "secuestro": secuestro,
    "terrorismo": terrorismo,
    "violencia_intrafamiliar": violencia_intrafamiliar
}
```

### Aquí unifico el formato de fecha de cada uno de los dataframes

```python
del_med_ambiente["FECHA HECHO"] = pd.to_datetime(
    del_med_ambiente["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

del_informaticos["FECHA HECHO"] = pd.to_datetime(
    del_informaticos["FECHA HECHO"], 
    format='%Y %b %d %I:%M:%S %p', 
    errors='coerce') # Acá el formato es ligereamente diferente porque el dataframe original estaba medio raro

del_sexuales["FECHA HECHO"] = pd.to_datetime(
    del_sexuales["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

extorsion["FECHA HECHO"] = pd.to_datetime(
    extorsion["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

hurto_vehiculos["FECHA HECHO"] = pd.to_datetime(
    hurto_vehiculos["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

hurto_personas["FECHA HECHO"] = pd.to_datetime(
    hurto_personas["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

incautacion_estupefacientes["FECHA HECHO"] = pd.to_datetime(
    incautacion_estupefacientes["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

masacres["FECHA HECHO"] = pd.to_datetime(
    masacres["FECHA HECHO"], 
    format='%Y %b %d %I:%M:%S %p', 
    errors='coerce') # Acá el formato es ligereamente diferente porque el dataframe original estaba medio raro

pirateria_terrestre["FECHA HECHO"] = pd.to_datetime(
    pirateria_terrestre["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

secuestro["FECHA HECHO"] = pd.to_datetime(
    secuestro["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

terrorismo["FECHA HECHO"] = pd.to_datetime(
    terrorismo["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')

violencia_intrafamiliar["FECHA HECHO"] = pd.to_datetime(
    violencia_intrafamiliar["FECHA HECHO"], 
    format='mixed', 
    errors='coerce')
```

### Acá unifico el formato de nombres para cada departamento ya que se suplicaban los departamentos en el merge de abajo. 

```python
for nombre, df in datasets.items():
    df["DEPARTAMENTO"] = (df["DEPARTAMENTO"]
                          .str.replace("Á", "A")
                          .str.replace("É", "E")
                          .str.replace("Í", "I")
                          .str.replace("Ó", "O")
                          .str.replace("Ú", "U")
                          .str.replace("SAN ANDRES ISLAS", "SAN ANDRES")
                          .str.replace("VALLE", "VALLE DEL CAUCA")
                          .str.replace("BOGOTA D.C.", "BOGOTA")
                          .str.replace("LA GUAJIRA", "GUAJIRA")
                          .str.replace("VALLE DEL CAUCA DEL CAUCA", "VALLE DEL CAUCA")
                        )
    
```

---

### __(Leo) Crear la columna año y eliminar filas fuera del rango__

```python
for nombre, df in datasets.items():
    df["ANO"] = df["FECHA HECHO"].dt.year
    
for nombre, df in datasets.items():
    print(f"\nDataframe: {nombre}, shape: {df.shape}")
    print("Año mínimo:", df["ANO"].min())
    print("Año máximo:", df["ANO"].max())
    print("="*50)
```

```python
for nombre, df in datasets.items():
    datasets[nombre] = df[(df["ANO"] >= 2023) & (df["ANO"] <= 2025)]

for nombre, df in datasets.items():
    print(f"{nombre}: {df.shape}") # Para conocer en cuanto se reduce el tamaño de los dataframes luego de filtrar por el rango de años
```

---

### __(Leo) Revisión de los valores nulos de columnas por dataset__

Aquí reviso para todos los datasets la suma de valores nulos que tienen cada una de sus columnas

```python
for nombre, df in datasets.items():
    print(f"\nDataset: {nombre}")
    print(df.isnull().sum().sort_values(ascending=False).to_string())
```

Estos son mis hallazgos:
1. El dataset de delitos sexuales tiene 353 valores nulos para la columna de SEXO. La cantidad de filas que tiene el dataframe es de 451146. Como en comparación los nulos son insignificantes decido eliminarlos. 
3. El dataset de masacres tiene 359 valores nulos para la columna ZONA. La cantidad de filas que tiee el dataframe es de 451, por esta razón intuyo que sencillamente no se implementó conocer la zona exacta de la masacre y decido eliminar la columna por completo. 

```python
del_sexuales.dropna(inplace=True) 
print(f"Shape de delitos sexuales luego de hacer dropna: {del_sexuales.shape}")

masacres.drop(columns=["ZONA"], inplace=True, errors="ignore")
print(f"Shape de masacres luego de eliminar columna ZONA: {masacres.shape}")
```

---

### __(Leo) Revisión de valores duplicados en columnas por dataset__

Aquí reviso para todas las columnas de todos los datasets la suma de valores repetidos

```python
for nombre, df in datasets.items():
    print(f"\nDataframe: {nombre}, shape: {df.shape}")
    print("Número de duplicados:",df.duplicated(keep="first").sum()) # El parámetro keep="first" indica que se mantendrá la primera aparición de cada fila duplicada y se marcarán como duplicadas las siguientes apariciones.
    print("="*50)
```

Para este punto de la limpieza de datos decidí dejar los registros intactos en cuanto a duplicados debido a que el hecho de que exista un duplicado no representa automaticamente un error, lo que representa es que en el mismo día y en el mismo lugar sucedieron N cantidad de casos que pandas toma como duplicados debido a la falta de una columna que sirva para  como identificador único. 

---

### (Leo) Merge final de todos los datasets

```python
for nombre, df in datasets.items():
    print(f"Número de columnas = {len(df.columns.tolist())} | {nombre} | {df.columns.tolist()}")
    print("="*150)
```

```python
for nombre, df in datasets.items():
    print(f"{nombre}: {df.shape}")
```

```python
for nombre, df in datasets.items():
    d = df.copy()
    if nombre == "masacres":
        d = d.rename(columns={"CASOS": "CANTIDAD"})
    d = d.groupby(["DEPARTAMENTO", "ANO"], as_index=False)["CANTIDAD"].sum()
    d = d.rename(columns={"CANTIDAD": f"CANTIDAD_{nombre}"})
    datasets[nombre] = d
    
for nombre, df in datasets.items():
    print(nombre, df.shape)
```

```python
claves = ["DEPARTAMENTO", "ANO"]
df_final = datasets["del_med_ambiente"].merge(
    datasets["del_informaticos"], on=claves, how="outer"
    ).merge(
        datasets["del_sexuales"], on=claves, how="outer"
    ).merge(
        datasets["extorsion"], on=claves, how="outer"
    ).merge(
        datasets["hurto_vehiculos"], on=claves, how="outer"
    ).merge(
        datasets["hurto_personas"], on=claves, how="outer"
    ).merge(
        datasets["incautacion_estupefacientes"], on=claves, how="outer"
    ).merge(
        datasets["masacres"], on=claves, how="outer"
    ).merge(
        datasets["pirateria_terrestre"], on=claves, how="outer"
    ).merge(
        datasets["secuestro"], on=claves, how="outer"
    ).merge(
        datasets["terrorismo"], on=claves, how="outer"
    ).merge(
        datasets["violencia_intrafamiliar"], on=claves, how="outer"
    )
```

### EL RESULTADO FINAL DEL MERGE Y DE LA UNIFICACION DE DEPARTAMENTOS ANTERIOR ES EL SIGUIENTE:

```python
print("SHAPE FINAL DEL MERGE:", df_final.shape)
print("NUMERO DE DEPARTAMENTOS (CONTANDO A BOGOTA COMO UNO):", df_final["DEPARTAMENTO"].nunique())
```

### ENVIAR EL NUEVO CSV A LA CARPETA DE DATA PROCESSED:

```python
df_final.to_csv(carpeta_salida_csv / "Metricas delitos segun departamento - Filtrado 2023-2026.csv", index=False)
```
